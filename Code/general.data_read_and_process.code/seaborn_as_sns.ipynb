# 计数图的绘制

```python
%config InlineBackend.figure_format = 'svg'
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from matfonter import matfonter
FONT_SOURCE = "../../Assets/FontSourcesMacOS"
sns.set_style('white')
matfonter(FONT_SOURCE + '/CN_SF_ZCOOL_XiaoWei/ZCOOLXiaoWei-Regular.ttf')
tips = sns.load_dataset('tips')
data = {'Thur': '星期四', 'Fri': '星期五', 'Sat':'星期六', 'Sun': '星期日'}
tips['day'] = tips['day'].map(data)
ax = sns.countplot(x='day', data=tips, hue='day', palette='pastel',
              order=['星期四', '星期五', '星期六', '星期日'], legend=False, dodge=False)
ax.bar_label(ax.containers[0])

plt.title('小费计数图')
plt.xlabel('星期')
plt.ylabel('数量')
```
