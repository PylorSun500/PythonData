```python
ymax = df['最高气温']
ymin = df['最低气温']
x = np.arange(len(ymax))
# plt.plot(x, ymax, x, ymin) 一次性绘制多条
plt.plot(x, ymax, color='gray', linestyle = ':', marker = '3')
plt.plot(x, ymin, color='blue', ls='--', marker='^')
plt.title("北京市未来 15 天气温走势图")
plt.legend(loc=2)
plt.xlabel('日期')
plt.ylabel('气温')
plt.axhline(y = 29.5, ls = ':')
# 指向性注释文本
plt.annotate(text = '平均温度', xy = (2, 29.5), xytext = (1.8, 30.7), arrowprops=dict(arrowstyle='->'))
# 网格
plt.grid(axis='y', lw = 1, ls = ':', alpha=0.25) # axis = x, y, both
for i, (j, y) in enumerate(zip(ymax, ymin)):
    plt.text(i-0.2, j+0.3, j)
    plt.text(i-0.2, y+0.3, y)
x_ticks = df['日期'].dt.strftime('%m-%d') 
plt.table(
    cellText=[list(ymax), list(ymin)], 
    rowLabels=['最高气温', '最低气温'], 
    colLabels=list(x_ticks[0]), 
    loc='best', 
    fontsize=32, 
    colWidths=[0.05]*15
)
```

```python
plt.rcParams['axes.unicode_minus'] = False
xrec = np.linspace(-np.pi, np.pi, 1000)
ysinx = np.sin(xrec)
plt.plot(xrec, ysinx)
plt.annotate(text = '平均温度', xy = (2, -1), xytext = (1.8, -1), arrowprops=dict(arrowstyle='->'))
```
