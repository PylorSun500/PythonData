```python
%matplotlib inline
%config InlineBackend.figure_format = 'svg'

from matfonter import matfonter
import matplotlib.pyplot as plt

matfonter("/Users/pylorsun/Documents/Study/2026term2/PythonData/Assets/FontSourcesMacOS/CN_SA_OPPO_Sans_4.0/OPPO Sans 4.0.ttf")

times = ['00:00', '02:00', '04:00', '06:00', '08:00', '10:00',
         '12:00', '14:00', '16:00', '18:00', '20:00', '22:00']
wind_speed = [8, 9, 11, 14, 8, 15, 22, 11.5, 10.5, 11.5, 11.5, 13.5]

fig, ax = plt.subplots(figsize=(8, 6))

ax.plot(times, wind_speed, marker='^', color='#1f77b4', linewidth=1.5,markerfacecolor='red',markeredgecolor='black')

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

for i, speed in enumerate(wind_speed):

    ax.text(x=i,y=speed + 0.3,s=str(speed),ha='center',va='bottom',fontsize=9,color='black')

ax.set_title('深圳市24小时的平均风速')
ax.set_xlabel('时间 (h)')
ax.set_ylabel('平均风速 (km/h)')

ax.set_ylim(7, 23)

plt.show()
```
