```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import matfonter
%matplotlib auto
FONT_SOURCE = '/Users/pylorsun/Documents/Study/2026term2/PythonData/Assets/FontSourcesMacOS/'
matfonter.matfonter(FONT_SOURCE + 'CN_SA_寒蝉端黑体 v1.3/ChillDuanSansVF.ttf')
fig = plt.figure(figsize=(12, 6))
ax = fig.add_subplot(111)

data = pd.read_excel('/Users/pylorsun/Documents/Study/2026term2/PythonData/Assets/data.xls', sheet_name='国内旅游情况')
subset = data.iloc[1:3, 1:]
x1 = np.arange(len(subset.columns))
bar_width = 0.4
ax.bar(x1, subset.iloc[0], width=bar_width, label='城镇居民')
ax.bar(x1 + bar_width, subset.iloc[1], width=bar_width, label = '农村居民游客')
ax.legend()
ax.set_title('游客情况')
ax.set_xlabel('年份')
ax.set_ylabel('人数')
ax.set_xticklabels(subset.columns, fontsize = 10)
```

```python
import seaborn as sns
DATASET_SOURCE = '/Users/pylorsun/Documents/Study/2026term2/PythonData/Assets/'
data = pd.read_excel(DATASET_SOURCE + '/data.xls', sheet_name='国内旅游情况')
subset = data.iloc[1:3, 1:]
df_long = subset.T.reset_index()
df_long.columns = ['年份', '城镇居民', '农村居民']
df_long = df_long.melt(id_vars='年份', var_name='游客类型', value_name='游客人数')
sns.barplot(data = df_long, x='年份', y='游客人数', hue='游客类型')
plt.xticks(fontsize = 8, rotation = 45)
```
